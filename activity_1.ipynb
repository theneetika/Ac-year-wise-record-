```python
df = pd.read_excel('//content/sample_sales_dataset (1).xlsx')

df.head()
```

```python
import pandas as pd
import plotly.express as px

# The DataFrame 'df' is already loaded from a previous cell.
# Convert 'Date' column to datetime objects to extract month
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.month_name()

sales_trend = df.groupby('Month')['Sales'].sum().reset_index()

# Ensure the months are in chronological order for plotting
month_order = [
    'January', 'February', 'March', 'April', 'May', 'June',
    'July', 'August', 'September', 'October', 'November', 'December'
]
sales_trend['Month'] = pd.Categorical(sales_trend['Month'], categories=month_order, ordered=True)
sales_trend = sales_trend.sort_values('Month')

fig = px.line(
    sales_trend,
    x='Month',
    y='Sales',
    title='Monthly Sales Trend'
)

fig.show()
```

```python
# Create a 'Profit' column for visualization purposes, assuming a 20% profit margin on sales.
df['Profit'] = df['Sales'] * 0.20

fig = px.scatter(
    df,
    x='Sales',
    y='Profit',
    title='Sales vs Profit',
    hover_data=['Product']
)

fig.show()
```

```python
fig = px.histogram(
    df,
    x='Sales',
    nbins=20,
    title='Sales Distribution'
)

fig.show()
```

```python
fig = px.sunburst(
    df,
    path=['Region','Product'],
    values='Sales',
    title='Sales Hierarchy'
)

fig.show()
```

```python
import plotly.graph_objects as go

# Group by 'Region' and 'Product' since 'Category' is not available
data = df.groupby(['Region','Product'])['Sales'].sum().reset_index()

# Create labels for nodes from unique regions and products
labels = list(pd.concat([data['Region'], data['Product']]).unique())

# Map regions and products to their index in the labels list
source = data['Region'].apply(lambda x: labels.index(x))
target = data['Product'].apply(lambda x: labels.index(x))
value = data['Sales']

fig = go.Figure(data=[go.Sankey(
    node=dict(label=labels),
    link=dict(
        source=source,
        target=target,
        value=value
    )
)])

fig.update_layout(title_text="Region to Product Sales Flow")

fig.show()
```

```python
fig = px.treemap(
    df,
    path=['Region','Product'],
    values='Sales',
    color='Profit',
    title='Sales Treemap'
)
fig.show()
```

```python
fig = px.box(
    df,
    x='Product',
    y='Sales',
    color='Product',
    title='Sales Distribution by Product'
)
fig.show()
```

```python
fig = px.area(df, x='Month', y='Sales')
fig.show()
```

```python
fig = px.violin(df, x='Product', y='Sales', box=True)
fig.show()
```

```python
import plotly.express as px

heat = df.pivot_table(
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum'
)

fig = px.imshow(heat, text_auto=True)
fig.show()
```
